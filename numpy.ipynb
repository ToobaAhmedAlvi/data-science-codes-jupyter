```python
import numpy as np
my_array = np.array([1,2,34,5])
x = np.array([1,4,5,6])
print("x:\t",x)
print('my_array:',my_array)
```

```python
print("Append array:\n",np.append(my_array,x))
```

```python
print("Concatenate:\n",np.concatenate((my_array,x)))
```

```python
print("Stack row Wise:")
print(np.vstack((my_array,x)))
```

```python
print("stack horizontally:")
print(np.hstack((my_array,x)))
```

```python
print("Another way of horizontally:")
print(np.r_[my_array,x])
```

```python
print("Stack column wise repeat:")
print(np.column_stack((my_array,x)))
print("column wise repeat:")
print(np.c_[my_array,x])
```

```python
 #Create an array
points = np.arange(-5, 5, 0.01)

# Make a meshgrid
xs, ys = np.meshgrid(points, points)
z = np.sqrt(xs ** 2 + ys ** 2)

# Display the image on the axes
plt.imshow(z, cmap=plt.cm.Reds)

# Draw a color bar
plt.colorbar()

# Show the plot
plt.show()
```
