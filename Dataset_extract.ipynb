```python
from google.colab import drive
import os
import zipfile
import shutil

# Mount Google Drive
drive.mount('/content/drive')
```

```python
# --- Configuration ---

# The full path to your zip file on Google Drive
# Assuming it's in a folder named 'Projects' in your Drive root:
zip_file_path = '/content/drive/MyDrive/PCB_Training_Results/teacher_pcb/YOLOv8l_pcb_teacher_compatible.zip'

# The directory where you want to extract contents locally in Colab's temporary storage
# Colab's temporary storage path is '/content/'
extract_dir_local = '/content/extracted_zip'

# The destination path on Google Drive where you want the new folder created
save_folder_drive = '/content/drive/MyDrive/Teacher_PCB'

# --- End Configuration ---
```

```python
# Ensure the local extraction directory exists
os.makedirs(extract_dir_local, exist_ok=True)

# 1. Extract the zip file locally
if os.path.exists(zip_file_path):
    try:
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(extract_dir_local)
        print(f"Files extracted locally to: {extract_dir_local}")

        # 2. Move the extracted contents from local storage to Google Drive
        # Create the destination folder on Drive if it doesn't exist
        os.makedirs(save_folder_drive, exist_ok=True)

        # We need to move the *contents* of the local extraction dir into the new drive folder
        # The shutil.move function works best for moving directories

        # Note: If the extraction creates a single folder inside 'extracted_zip', we move that inner folder.
        # If it extracts files directly, we move the contents.

        # A simple way to handle moving the entire result is using shutil.move
        # For simplicity here, we assume we move all individual items

        # Better approach: Use shutil.copytree or just ensure the output goes to the final drive path
        # Let's adjust the logic slightly to save directly to the Drive path if possible/practical

        # Re-extraction directly to the final Drive location might cause issues with large files or network latency.
        # The standard approach is local extraction then copying/moving.

        # We'll use shutil.copytree to ensure all files/subdirectories are moved correctly.

        # Ensure the destination folder is empty or doesn't exist for shutil.copytree to work correctly
        if os.path.exists(save_folder_drive):
             shutil.rmtree(save_folder_drive) # Remove existing folder first

        shutil.copytree(extract_dir_local, save_folder_drive)
        print(f"Contents moved successfully from local temp to Google Drive folder: {save_folder_drive}")

        # Optional: Clean up the local extracted files
        shutil.rmtree(extract_dir_local)
        print(f"Cleaned up local temporary files in: {extract_dir_local}")


    except FileNotFoundError:
        print(f"Error: Zip file not found at '{zip_file_path}'. Check your path and file name.")
    except zipfile.BadZipFile:
        print(f"Error: The file at '{zip_file_path}' is not a valid zip file or is corrupted.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
else:
    print(f"Zip file not found at specified path: {zip_file_path}")

```
