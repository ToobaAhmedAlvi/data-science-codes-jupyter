```python
import pandas as pd
# Assume 'path' holds the dataset directory from kagglehub.dataset_download()
df = pd.read_csv(f'{path}/train.csv') # Or adjust path/filename
df['Full_Text'] = df['Title'].astype(str) + " " + df['Description'].astype(str)
documents = df['Full_Text'].tolist()
ground_truth_labels = df['Class Index'].values # Used for NMI & Accuracy
K = 4 # AG News has 4 classes
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load a powerful pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

def get_embeddings(texts):
    """Converts a list of texts into a numpy array of embeddings."""
    # SBERT handles the tokenization and vector generation
    embeddings = model.encode(texts, convert_to_numpy=True)
    return embeddings

# 💡 Generate the document embeddings (X)
X = get_embeddings(documents)
```

```python
from sklearn.cluster import KMeans

# 💡 K-Means (The 'vanilla' version)
kmeans_vanilla = KMeans(
    n_clusters=K,
    init='k-means++',      # Use k-means++ for good initialization
    n_init='auto',         # Run multiple times to find the best result
    random_state=42,
    max_iter=100
).fit(X)

# Results
vanilla_labels = kmeans_vanilla.labels_
vanilla_centroids = kmeans_vanilla.cluster_centers_
```

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.nlp.stemmers import Stemmer
from sumy.utils import get_stop_words
import nltk
# Download resources if not already present
# nltk.download('punkt')

def get_cluster_summary(docs_list, sentences_count=3):
    """Generates an extractive summary for a list of documents."""
    # Concatenate all cluster documents into one string
    full_text = ' '.join(docs_list)
    
    # Set up TextRank components
    parser = PlaintextParser.from_string(full_text, Tokenizer("english"))
    stemmer = Stemmer("english")
    summarizer = TextRankSummarizer(stemmer)
    summarizer.stop_words = get_stop_words("english")
    
    # Generate summary
    summary = " ".join([str(s) for s in summarizer(parser.document, sentences_count)])
    return summary
```

## K-NLP Means Customized Algorithm

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import normalized_mutual_info_score, accuracy_score
# 🛠️ FIX: Correct import for the Hungarian algorithm (Linear Assignment)
from scipy.optimize import linear_sum_assignment 


# 🛠️ HELPER FUNCTION: To correctly calculate Clustering Accuracy
def map_labels_to_ground_truth(y_true, y_pred):
    """
    Maps predicted cluster labels to true labels to maximize accuracy.
    Uses the Hungarian algorithm (linear_sum_assignment).
    """
    labels = np.unique(y_pred)
    true_labels = np.unique(y_true)
    
    # Initialize cost matrix (K x K, where K is number of clusters)
    # The cost is the negative count of agreements (to maximize agreement)
    cost_matrix = np.zeros((labels.size, true_labels.size), dtype=np.int64)
    for i, pred_label in enumerate(labels):
        for j, true_label in enumerate(true_labels):
            # Count the number of agreements
            agreement_count = np.sum((y_pred == pred_label) & (y_true == true_label))
            # Set the cost to negative of agreement count for minimization
            cost_matrix[i, j] = -agreement_count

    # Find the optimal assignment (row_ind is predicted label index, col_ind is true label index)
    row_ind, col_ind = linear_sum_assignment(cost_matrix)
    
    # Create a mapping from predicted labels to true labels
    label_mapping = {labels[i]: true_labels[j] for i, j in zip(row_ind, col_ind)}
    
    # Apply the mapping
    mapped_y_pred = np.array([label_mapping.get(label, label) for label in y_pred])
    
    return mapped_y_pred


# --- NOTE: Dummy Functions for Standalone Code ---
# Replace these with your actual get_cluster_summary and get_embeddings functions
def get_cluster_summary(docs_list):
    """
    Placeholder for the TextRank/Summa implementation.
    """
    # Simulate an expensive operation
    import time
    # Time cost proportional to the number of sampled documents (N_SAMPLE)
    time.sleep(0.005 * len(docs_list)) 
    
    if len(docs_list) > 0:
        return f"Summary of {len(docs_list)} documents: {docs_list[0][:20]}..."
    return "Empty Summary"

def get_embeddings(texts):
    """
    Placeholder for the SBERT/Transformer embedding function.
    """
    # Simulate a heavy embedding operation
    import time
    time.sleep(0.1 * len(texts))  
    # Assuming embeddings have a dimension of 768
    return [np.random.rand(768) for _ in texts]


def k_nlp_means_faster(X, documents, K, T=100, l=5, N_SAMPLE=100):
    """
    Custom K-NLPMeans implementation with Centroid-Based Document Sampling for speedup.
    
    :param X: Document embeddings (N x D array).
    :param documents: List of raw document strings (N elements).
    :param K: Number of clusters.
    :param T: Max iterations.
    :param l: Centroid update interval.
    :param N_SAMPLE: The number of documents to sample from a cluster for summarization.
    :return: final_labels, final_summaries
    """
    
    # 1. Initialization (using k-means++ for initial centroids)
    kmeans_init = KMeans(n_clusters=K, init='k-means++', n_init=1, random_state=42, max_iter=1).fit(X)
    centroids = kmeans_init.cluster_centers_
    
    print(f"Starting K-NLPMeans with K={K}, T={T}, l={l}, and N_SAMPLE={N_SAMPLE} for summarization.")
    
    # A list to hold document strings grouped by cluster (only prepared when needed)
    
    for t in range(1, T + 1):
        # 2. Assignment Step: Assign each document to the nearest centroid
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)
        
        # 3. Update Step
        new_centroids = np.zeros_like(centroids)
        
        if t % l != 0:
            # Standard K-Means Step (fast numerical average)
            for j in range(K):
                cluster_points = X[labels == j]
                if cluster_points.size > 0:
                    new_centroids[j] = np.mean(cluster_points, axis=0)
                else:
                    new_centroids[j] = centroids[j]
        
        else:
            # Summarization Step (expensive, done every 'l' iterations)
            print(f"--- Iteration {t}: Running Centroid-Based Document Sampling Update (N={N_SAMPLE}) ---")
            
            for j in range(K):
                cluster_indices = np.where(labels == j)[0]
                
                if cluster_indices.size > 0:
                    
                    # Centroid-Based Sampling (Speed-up applied here)
                    cluster_embeddings = X[cluster_indices]
                    distances_to_centroid = np.linalg.norm(cluster_embeddings - centroids[j], axis=1)
                    closest_doc_local_indices = np.argsort(distances_to_centroid)[:N_SAMPLE]
                    
                    # Use the documents list for summarization
                    # Note: We index the main 'documents' list using the indices from X
                    docs_to_summarize = [documents[i] for i in cluster_indices[closest_doc_local_indices]]
                    
                    # Textual Summarization and Embedding
                    summary = get_cluster_summary(docs_to_summarize)
                    new_centroids[j] = get_embeddings([summary])[0]
                else:
                    new_centroids[j] = centroids[j]
                    
        # Check for convergence
        if np.linalg.norm(new_centroids - centroids) < 1e-4:
            print(f"Converged at iteration {t}")
            break
            
        centroids = new_centroids

    final_labels = labels
    final_centroids = centroids
    
    # Final textual summaries (using the same sampling for speed/consistency)
    final_summaries = []
        
    for j in range(K):
        cluster_indices = np.where(final_labels == j)[0]
        if cluster_indices.size > 0:
            cluster_embeddings = X[cluster_indices]
            distances_to_centroid = np.linalg.norm(cluster_embeddings - final_centroids[j], axis=1)
            closest_doc_local_indices = np.argsort(distances_to_centroid)[:N_SAMPLE]
            docs_to_summarize = [documents[i] for i in cluster_indices[closest_doc_local_indices]]
            final_summaries.append(get_cluster_summary(docs_to_summarize))
        else:
            final_summaries.append("Empty Cluster")


    return final_labels, final_summaries
```

```python
# --- RUN THE CLUSTERING ALGORITHM ---
# IMPORTANT: Ensure X (embeddings), documents (raw text), and K_CLUSTERS are defined
# in a cell *before* this one.

K_CLUSTERS = 4 # Use the number of clusters you are working with

# This line runs your algorithm and DEFINES the 'nlp_labels' variable
# If this line fails or is skipped, the NameError occurs.
nlp_labels, nlp_summaries = k_nlp_means_faster(
    X, 
    documents, 
    K=K_CLUSTERS, 
    T=75,          # Example max iterations
    l=10,          # Example update interval
    N_SAMPLE=100   # Example sample size
)

print("K-NLPMeans finished. The 'nlp_labels' variable is now defined.")
```

## Comparsion of Performance Using ACC and NMI

```python
from sklearn.metrics import normalized_mutual_info_score, accuracy_score
# Change this import:
from scipy.optimize import linear_sum_assignment # Replaces sklearn.utils.linear_assignment_

# Rest of your code:
# The function you use will also need to change from linear_assignment to linear_sum_assignment
# E.g., old_code[row_ind, col_ind] = value becomes new_code[row_ind, col_ind] = value
# A. NMI (Label-independent)
nmi_vanilla = normalized_mutual_info_score(ground_truth_labels, vanilla_labels)
nmi_nlp = normalized_mutual_info_score(ground_truth_labels, nlp_labels)
print(f"K-Means NMI: {nmi_vanilla:.4f}")
print(f"K-NLPMeans NMI: {nmi_nlp:.4f}")

# B. Accuracy (Requires optimal mapping using Hungarian Algorithm)
def cluster_acc(y_true, y_pred):
    """Calculates clustering accuracy by finding the optimal label mapping."""
    # Create an assignment matrix
    D = max(y_pred.max(), y_true.max()) + 1
    w = np.zeros((D, D), dtype=int)
    for i in range(y_pred.size):
        w[y_pred[i], y_true[i]] += 1
        
    # Find the optimal one-to-one mapping (maximum weight matching)
    row_ind, col_ind = linear_sum_assignment(-w)
    
    # Return accuracy
    return w[row_ind, col_ind].sum() * 1.0 / y_pred.size

acc_vanilla = cluster_acc(ground_truth_labels, vanilla_labels)
acc_nlp = cluster_acc(ground_truth_labels, nlp_labels)
print(f"K-Means Accuracy: {acc_vanilla:.4f}")
print(f"K-NLPMeans Accuracy: {acc_nlp:.4f}")
```
