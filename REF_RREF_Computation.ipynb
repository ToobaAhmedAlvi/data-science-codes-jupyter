```python
import sympy as sp

# Define a matrix
A = sp.Matrix([
    [2, 4, -2, 2,3],
    [4, 9, -3, 8,6],
    [-2, -3, 7, 10,7]
])

# Compute the Row Echelon Form (REF)
echelon_matrix = A.echelon_form()

# Print the REF matrix
print("Echelon Form (REF) Matrix:")
sp.pprint(echelon_matrix)
```

```python
import sympy as sp

# Define a matrix
A = sp.Matrix([
    [2, 4, -2, 2,3],
    [4, 9, -3, 8,6],
    [-2, -3, 7, 10,7]
])

# Compute the Reduced Row Echelon Form (RREF)
rref_matrix, pivot_columns = A.rref()

# Print the RREF matrix
print("RREF Matrix:")
sp.pprint(rref_matrix)

# Print pivot columns
print("\nPivot Columns:", pivot_columns)
```
