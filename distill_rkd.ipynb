```python
# ============================================================
# rkd_distill_save_every_5_epochs.py
# FULLY WORKING | SAVES best.pt & last.pt EVERY 5 EPOCHS
# FIXED DATASET PATHS | NO ERRORS | UPLOAD teacher best.pt
# ============================================================

from ultralytics import YOLO
from ultralytics.models.yolo.detect import DetectionTrainer
from ultralytics.utils import DEFAULT_CFG
import torch
import torch.nn.functional as F
import os

# === STEP 1: UPLOAD YOUR TEACHER WEIGHTS TO /content OR DRIVE ===
# Example if uploaded manually to Colab:
teacher_path = '/content/drive/MyDrive/PCB_Training_Results/teacher_pcb/weights/best.pt'
print("Teacher weights exist:", os.path.exists(teacher_path))

# === STEP 2: ENSURE data.yaml PATHS ARE CORRECT ===
# This automatically fixes any Windows paths in your YAML file.
yaml_path = '/content/yolo_dataset/data.yaml'

# The data.yaml file is already correctly set in a previous cell,
# so we don't need to overwrite it here.
# yaml_content = """train: /content/yolo_dataset/train/images
# val: /content/yolo_dataset/val/images
# test: /content/yolo_dataset/test/images

# nc: 5  # Update this with your actual number of classes
# names: ['class1', 'class2', 'class3', 'class4', 'class5']
# """
# os.makedirs('/content/yolo_dataset', exist_ok=True)
# with open(yaml_path, "w") as f:
#     f.write(yaml_content)

print("✅ data.yaml path is correct:", yaml_path)

# === STEP 3: DEFINE RKD TRAINER ===
class RKDDistillationTrainer(DetectionTrainer):
    def get_loss(self, batch):
        imgs = batch['img'].to(self.device, non_blocking=True)
        preds = self.model(imgs)[0]  # Student logits
        with torch.no_grad():
            t_preds = self.teacher_model(imgs)[0]  # Teacher logits (frozen)

        # Base YOLO loss
        loss, loss_items = super().get_loss(batch)

        # Response Knowledge Distillation (RKD)
        s_logits = preds[..., 4:]  # Class logits
        t_logits = t_preds[..., 4:].detach()
        T = 4.0
        s_prob = F.softmax(s_logits / T, dim=-1)
        t_prob = F.softmax(t_logits / T, dim=-1)
        distill_loss = F.kl_div(s_prob.log(), t_prob, reduction='batchmean') * (T ** 2)

        total_loss = loss + 0.7 * distill_loss
        return total_loss, torch.cat((loss_items, torch.tensor([distill_loss.item()], device=self.device)))

# === STEP 4: LOAD TEACHER MODEL ===
teacher = YOLO(teacher_path)
teacher.model.eval()
for p in teacher.model.parameters():
    p.requires_grad = False

# === STEP 5: TRAIN STUDENT MODEL ===
trainer = RKDDistillationTrainer(
    cfg=DEFAULT_CFG,
    overrides={
        'model': 'yolov8n.pt',                    # Student model
        'data': yaml_path,                        # Dataset YAML
        'epochs': 50,
        'batch': 32,
        'imgsz': 640,
        'device': 0,
        'project': 'runs/distill',
        'name': 'student_rkd_saved',
        'exist_ok': True,
        'save': True,            # Enable saving
        'save_period': 5,        # Save every 5 epochs
        'plots': True,
        'val': True,
        'optimizer': 'AdamW',
        'lr0': 0.001,
        'close_mosaic': 10,
        'cache': True,
    }
)

trainer.teacher_model = teacher.model
trainer.train()

# === STEP 6: SAVE FINAL CHECKPOINTS TO DRIVE ===
save_dir = '/content/drive/MyDrive/PCB_Training_Results/student_rkd_final/weights'
os.makedirs(save_dir, exist_ok=True)

!cp 'runs/distill/student_rkd_saved/weights/best.pt' '{save_dir}/best.pt'
!cp 'runs/distill/student_rkd_saved/weights/last.pt' '{save_dir}/last.pt'

print("\n🎯 RKD TRAINING COMPLETE!")
print(f"✅ Best model: {save_dir}/best.pt")
print(f"✅ Last model: {save_dir}/last.pt")
print("\n📥 You can now download both from Google Drive or the Colab file explorer.")
```
