```python
#!/usr/bin/env python3
"""
Training a Foreground Segmentation Tool with DINOv3

This script trains a linear foreground segmentation model using DINOv3 features.
"""

import os
import pickle
import datetime

from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import average_precision_score
from sklearn.linear_model import LogisticRegression
import torch
import torchvision.transforms.functional as TF
from tqdm import tqdm

def dice_score(y_true, y_pred):
    """Calculate dice score for binary segmentation."""
    y_true = y_true.astype(bool)
    y_pred = y_pred.astype(bool)
    
    intersection = (y_true & y_pred).sum()
    total = y_true.sum() + y_pred.sum()
    
    if total == 0:
        return 1.0  # Perfect score when both are empty
    
    return 2.0 * intersection / total

def main():
    print("=== Training a Foreground Segmentation Tool with DINOv3 ===")
    
    # Create timestamped output directory
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    output_dir = f"segmentation_results_{timestamp}"
    os.makedirs(output_dir, exist_ok=True)
    print(f"Results will be saved to: {output_dir}")
    
    # Global constants
    DINOV3_GITHUB_LOCATION = "facebookresearch/dinov3"
    
    if os.getenv("DINOV3_LOCATION") is not None:
        DINOV3_LOCATION = os.getenv("DINOV3_LOCATION")
    else:
        DINOV3_LOCATION = DINOV3_GITHUB_LOCATION
    
    # Model constants
    MODEL_DINOV3_VITS = "dinov3_vits16"
    MODEL_DINOV3_VITSP = "dinov3_vits16plus"
    MODEL_DINOV3_VITB = "dinov3_vitb16"
    MODEL_DINOV3_VITL = "dinov3_vitl16"
    MODEL_DINOV3_VITHP = "dinov3_vith16plus"
    MODEL_DINOV3_VIT7B = "dinov3_vit7b16"
    
    MODEL_NAME = MODEL_DINOV3_VITL
    
    # Data constants
    PATCH_SIZE = 16
    IMAGE_SIZE = 768
    IMAGENET_MEAN = (0.485, 0.456, 0.406)
    IMAGENET_STD = (0.229, 0.224, 0.225)
    
    MODEL_TO_NUM_LAYERS = {
        MODEL_DINOV3_VITS: 12,
        MODEL_DINOV3_VITSP: 12,
        MODEL_DINOV3_VITB: 12,
        MODEL_DINOV3_VITL: 24,
        MODEL_DINOV3_VITHP: 32,
        MODEL_DINOV3_VIT7B: 40,
    }
    
    n_layers = MODEL_TO_NUM_LAYERS[MODEL_NAME]
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    
    print(f"DINOv3 location set to {DINOV3_LOCATION}")
    print(f"Using device: {device}")
    print(f"Model: {MODEL_NAME} with {n_layers} layers")
    
    # Load DINOv3 model
    print("\\n1. Loading DINOv3 model...")
    model = torch.hub.load(
        repo_or_dir=DINOV3_LOCATION,
        model=MODEL_NAME,
        source="local" if DINOV3_LOCATION != DINOV3_GITHUB_LOCATION else "github",
    )
    
    model.to(device)
    model.eval()
    print(f"Model loaded successfully on {device}")
    
    # Data loading
    print("\\n2. Loading training data...")
    DATA_DIR = "data"
    IMAGES_DIR = os.path.join(DATA_DIR, "Image")
    MASKS_DIR = os.path.join(DATA_DIR, "Mask")
    
    def get_image_file_paths(folder_path: str) -> list:
        """Get sorted list of image file paths instead of loading all images."""
        image_files = sorted([f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
        return [os.path.join(folder_path, f) for f in image_files]
    
    def load_image_on_demand(image_path: str):
        """Load image only when needed."""
        try:
            return Image.open(image_path)
        except Exception as e:
            print(f"Warning: Could not load {image_path}: {e}")
            return None
        
    image_paths = get_image_file_paths(IMAGES_DIR)
    mask_paths = get_image_file_paths(MASKS_DIR)
    
    # Use only first 100 images for faster training and testing
    MAX_IMAGES = 100
    image_paths = image_paths[:MAX_IMAGES]
    mask_paths = mask_paths[:MAX_IMAGES]
    
    n_images = len(image_paths)
    assert n_images == len(mask_paths), f"{len(image_paths)=}, {len(mask_paths)=}"
    
    print(f"Loaded {n_images} images and labels")
    
    # Setup patch processing
    print("\\n3. Setting up patch processing...")
    patch_quant_filter = torch.nn.Conv2d(1, 1, PATCH_SIZE, stride=PATCH_SIZE, bias=False)
    patch_quant_filter.weight.data.fill_(1.0 / (PATCH_SIZE * PATCH_SIZE))
    
    def resize_transform(
        mask_image: Image.Image,
        image_size: int = IMAGE_SIZE,
        patch_size: int = PATCH_SIZE,
    ) -> torch.Tensor:
        w, h = mask_image.size
        h_patches = int(image_size / patch_size)
        w_patches = int((w * image_size) / (h * patch_size))
        return TF.to_tensor(TF.resize(mask_image, (h_patches * patch_size, w_patches * patch_size)))
    
    print("Patch processing setup complete")
    
    # Check if features already exist
    features_file = f"extracted_features_{timestamp}.npz"
    features_path = os.path.join(output_dir, features_file)
    
    # Look for existing feature files in current directory
    existing_features = [f for f in os.listdir('.') if f.startswith('segmentation_results_') and os.path.isdir(f)]
    load_existing = False
    
    if existing_features:
        latest_dir = max(existing_features, key=os.path.getctime)
        existing_feature_files = [f for f in os.listdir(latest_dir) if f.startswith('extracted_features_') and f.endswith('.npz')]
        if existing_feature_files:
            existing_path = os.path.join(latest_dir, existing_feature_files[0])
            print(f"Found existing features at: {existing_path}")
            response = input("Load existing features? (y/n): ").lower().strip()
            if response == 'y':
                load_existing = True
                features_path = existing_path
    
    if load_existing:
        print(f"Loading existing features from {features_path}...")
        data = np.load(features_path)
        xs = torch.from_numpy(data['xs'])
        ys = torch.from_numpy(data['ys'])
        image_index = torch.from_numpy(data['image_index'])
        print("Features loaded successfully!")
    else:
        # Feature extraction
        print("\\n4. Extracting features from all images...")
        xs = []
        ys = []
        image_index = []
        
        with torch.inference_mode():        
            for i in tqdm(range(n_images), desc="Processing images"):
                # Loading the ground truth (convert to grayscale for binary masks)
                mask_i = load_image_on_demand(mask_paths[i])
                if mask_i is None:
                    continue
                mask_i = mask_i.convert('L')
                mask_i_resized = resize_transform(mask_i)
                mask_i_quantized = patch_quant_filter(mask_i_resized).squeeze().view(-1).detach().cpu()
                ys.append(mask_i_quantized)
                
                # Loading the image data 
                image_i = load_image_on_demand(image_paths[i])
                if image_i is None:
                    continue
                image_i = image_i.convert('RGB')
                image_i_resized = resize_transform(image_i)
                image_i_resized = TF.normalize(image_i_resized, mean=IMAGENET_MEAN, std=IMAGENET_STD)
                image_i_resized = image_i_resized.unsqueeze(0).to(device)
        
                feats = model.get_intermediate_layers(image_i_resized, n=range(n_layers), reshape=True, norm=True)
                dim = feats[-1].shape[1]
                xs.append(feats[-1].squeeze().view(dim, -1).permute(1,0).detach().cpu())
        
                image_index.append(i * torch.ones(ys[-1].shape))
        
        # Concatenate all lists into torch tensors 
        xs = torch.cat(xs)
        ys = torch.cat(ys)
        image_index = torch.cat(image_index)
        
        # keeping only the patches that have clear positive or negative label
        idx = (ys < 0.01) | (ys > 0.99)
        xs = xs[idx]
        ys = ys[idx]
        image_index = image_index[idx]
    
    print("Design matrix of size : ", xs.shape)
    print("Label matrix of size : ", ys.shape)
    
    # Save extracted features to avoid re-computation
    if not load_existing:
        features_path = os.path.join(output_dir, f"extracted_features_{timestamp}.npz")
        print(f"Saving extracted features to {features_path}...")
        np.savez_compressed(features_path, 
                           xs=xs.numpy(), 
                           ys=ys.numpy(), 
                           image_index=image_index.numpy())
        print("Features saved successfully!")
        print("You can now use 'python train_from_features.py' to continue training from these saved features.")
    
    # 5-fold Cross-validation for model selection
    print("\\n5. 5-fold Cross-validation for model selection...")
    from sklearn.model_selection import KFold
    
    cs = np.logspace(-7, 0, 8)
    n_folds = 5
    kfold = KFold(n_splits=n_folds, shuffle=True, random_state=42)
    
    ap_scores = np.zeros((n_folds, len(cs)))
    dice_scores = np.zeros((n_folds, len(cs)))
    
    # Convert image_index to numpy for sklearn compatibility
    image_indices = image_index.numpy()
    unique_images = np.unique(image_indices)
    
    # Create folds based on images (not patches)
    fold_idx = 0
    for train_imgs, val_imgs in kfold.split(unique_images):
        train_image_set = set(unique_images[train_imgs])
        val_image_set = set(unique_images[val_imgs])
        
        print(f'Fold {fold_idx + 1}/5: Train on {len(train_image_set)} images, validate on {len(val_image_set)} images')
        
        # Select patches based on image membership
        train_selection = np.array([img_idx in train_image_set for img_idx in image_indices])
        val_selection = np.array([img_idx in val_image_set for img_idx in image_indices])
        
        fold_x_train = xs[train_selection].numpy()
        fold_y_train = (ys[train_selection] > 0).long().numpy()
        fold_x_val = xs[val_selection].numpy()
        fold_y_val = (ys[val_selection] > 0).long().numpy()
    
        for j, c in enumerate(cs):
            print(f"  Training logistic regression with C={c:.2e}")
            clf = LogisticRegression(random_state=0, C=c, max_iter=10000).fit(fold_x_train, fold_y_train)
            output = clf.predict_proba(fold_x_val)
            predictions = clf.predict(fold_x_val)
            
            ap_s = average_precision_score(fold_y_val, output[:, 1])
            dice_s = dice_score(fold_y_val, predictions)
            
            ap_scores[fold_idx, j] = ap_s
            dice_scores[fold_idx, j] = dice_s
        
        fold_idx += 1
    
    print("Cross-validation complete!")
    
    # Find best C
    avg_ap_scores = ap_scores.mean(axis=0)
    avg_dice_scores = dice_scores.mean(axis=0)
    best_c_idx = np.argmax(avg_ap_scores)  # Select based on AP score
    best_c = cs[best_c_idx]
    
    print(f"\\nBest regularization C: {best_c:.1e}")
    print(f"Average precision: {avg_ap_scores[best_c_idx]:.3f}")
    print(f"Average dice score: {avg_dice_scores[best_c_idx]:.3f}")
    
    print("\\nAll cross-validation results:")
    print("C\t\tAP Score\tDice Score")
    print("-" * 35)
    
    # Save cross-validation results to file
    cv_results_path = os.path.join(output_dir, f"cv_results_{timestamp}.txt")
    with open(cv_results_path, "w") as f:
        f.write(f"Cross-Validation Results - {timestamp}\\n")
        f.write("="*50 + "\\n")
        f.write(f"Dataset: 100 images with 5-fold CV\\n")
        f.write(f"Best C: {best_c:.1e}\\n")
        f.write(f"Best Average Precision: {avg_ap_scores[best_c_idx]:.3f}\\n")
        f.write(f"Best Dice Score: {avg_dice_scores[best_c_idx]:.3f}\\n\\n")
        f.write("Detailed Results:\\n")
        f.write("C\t\tAP Score\tDice Score\\n")
        f.write("-" * 35 + "\\n")
        
        for j, c in enumerate(cs):
            result_line = f"{c:.2e}\t{avg_ap_scores[j]:.3f}\t\t{avg_dice_scores[j]:.3f}"
            print(result_line)
            f.write(result_line + "\\n")
    
    print(f"\\nCross-validation results saved to: {cv_results_path}")
    
    # Train final model
    print("\\n6. Training final model...")
    final_clf = LogisticRegression(random_state=0, C=best_c, max_iter=100000, verbose=1)
    final_clf.fit(xs.numpy(), (ys > 0).long().numpy())
    
    print(f"Final classifier trained with C={best_c:.1e}")
    
    # Test on new image (use first image from dataset as test)
    print("\\n7. Testing on sample image...")
    test_image = load_image_on_demand(image_paths[0])  # Use first image from dataset as test example
    test_image_resized = resize_transform(test_image)
    test_image_normalized = TF.normalize(test_image_resized, mean=IMAGENET_MEAN, std=IMAGENET_STD)
    
    with torch.inference_mode():
        feats = model.get_intermediate_layers(test_image_normalized.unsqueeze(0).to(device), n=range(n_layers), reshape=True, norm=True)
        x = feats[-1].squeeze().detach().cpu()
        dim = x.shape[0]
        x = x.view(dim, -1).permute(1, 0)
    
    h_patches, w_patches = [int(d / PATCH_SIZE) for d in test_image_resized.shape[1:]]
    
    fg_score = final_clf.predict_proba(x)[:, 1].reshape(h_patches, w_patches)
    fg_score_mf = signal.medfilt2d(fg_score, kernel_size=3)
    
    print("Test inference complete!")
    print(f"Foreground score shape: {fg_score.shape}")
    
    # Save visualization images
    print("\\nSaving result images...")
    
    # Save individual images with timestamps
    input_path = os.path.join(output_dir, f'test_image_input_{timestamp}.png')
    score_path = os.path.join(output_dir, f'test_image_foreground_score_{timestamp}.png') 
    filtered_path = os.path.join(output_dir, f'test_image_foreground_filtered_{timestamp}.png')
    
    plt.imsave(input_path, test_image_resized.permute(1, 2, 0).numpy())
    plt.imsave(score_path, fg_score, cmap='viridis')
    plt.imsave(filtered_path, fg_score_mf, cmap='viridis')
    
    # Create and save comparison visualization
    plt.figure(figsize=(12, 4), dpi=150)
    plt.subplot(1, 3, 1)
    plt.axis('off')
    plt.imshow(test_image_resized.permute(1, 2, 0))
    plt.title('Input Image', fontsize=14)
    
    plt.subplot(1, 3, 2)
    plt.axis('off')
    plt.imshow(fg_score, cmap='viridis')
    plt.title('Foreground Score', fontsize=14)
    plt.colorbar(shrink=0.8)
    
    plt.subplot(1, 3, 3)
    plt.axis('off')
    plt.imshow(fg_score_mf, cmap='viridis')
    plt.title('Filtered Score', fontsize=14)
    plt.colorbar(shrink=0.8)
    
    plt.tight_layout()
    comparison_path = os.path.join(output_dir, f'foreground_segmentation_results_{timestamp}.png')
    plt.savefig(comparison_path, dpi=150, bbox_inches='tight')
    plt.close()
    
    # Save training data sample visualization
    print("Saving training data sample...")
    data_index = 0
    image = load_image_on_demand(image_paths[data_index])
    mask = load_image_on_demand(mask_paths[data_index]).convert('L')  # Convert to grayscale
    
    # Create simple visualization for binary masks
    data_to_show = [image, mask]
    data_labels = ["Original", "Mask"]
    
    plt.figure(figsize=(8, 4), dpi=150)
    for i in range(len(data_to_show)):
        plt.subplot(1, len(data_to_show), i + 1)
        plt.imshow(data_to_show[i])
        plt.axis('off')
        plt.title(data_labels[i], fontsize=14)
    plt.tight_layout()
    training_sample_path = os.path.join(output_dir, f'training_data_sample_{timestamp}.png')
    plt.savefig(training_sample_path, dpi=150, bbox_inches='tight')
    plt.close()
    
    print("Images saved to directory:", output_dir)
    print(f"  - test_image_input_{timestamp}.png")
    print(f"  - test_image_foreground_score_{timestamp}.png") 
    print(f"  - test_image_foreground_filtered_{timestamp}.png")
    print(f"  - foreground_segmentation_results_{timestamp}.png")
    print(f"  - training_data_sample_{timestamp}.png")
    
    # Save model
    print("\\n8. Saving model...")
    model_path = os.path.join(output_dir, f"fg_classifier_{timestamp}.pkl")
    with open(model_path, "wb") as f:
        pickle.dump(final_clf, f)
        
    print(f"Classifier saved to {model_path}")
    print(f"Model performance: Average Precision = {avg_ap_scores[best_c_idx]:.3f}")
    print(f"Model performance: Dice Score = {avg_dice_scores[best_c_idx]:.3f}")
    print("\\n=== Foreground segmentation training completed successfully! ===")

if __name__ == "__main__":
    main()
```
