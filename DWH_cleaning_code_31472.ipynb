```python
#Checking for inconsistencies
def check_inconsistencies(df):
    # Check for missing values
    missing_values = df.isnull().sum()
    
    # Check for duplicate rows
    duplicate_rows = df.duplicated().sum()
    
    # Create a dictionary of inconsistencies
    inconsistencies = {
        "missing_values": missing_values,
        "duplicate_rows": duplicate_rows
    }
    
    return inconsistencies
```

### 4. Cleaning Ride Requests Dataset

```python
# Re-defined check_inconsistencies function
def check_inconsistencies(df):
    inconsistencies = {}

    # Check for invalid date formats
    for column in df.columns:
        if 'date' in column.lower() and df[column].dtype == 'object':  # Date columns
            try:
                pd.to_datetime(df[column], errors='raise')
            except Exception as e:
                inconsistencies[column] = "Invalid date format"
        
        # Check for negative or impossible values in numerical columns
        if df[column].dtype in ['float64', 'int64']:
            if df[column].min() < 0:
                inconsistencies[column] = "Negative values present"
        
        # Check for categorical columns with unexpected values
        if df[column].dtype == 'category':
            unique_values = df[column].unique()
            if len(unique_values) > 10:
                inconsistencies[column] = f"Too many unique values: {unique_values[:5]}"

    return inconsistencies
```

```python
# Loading ride requests dataset
ride_requests_df = pd.read_csv('C:\\Users\\ayesh\\Desktop\\DWH Assignments\\Dataset\\ride_requests.csv')
print(ride_requests_df)
```

```python
# Checking for missing values
ride_requests_missing = ride_requests_df.isnull().sum()
ride_requests_missing
```

```python
ride_requests_df['RequestedTime'] = ride_requests_df['RequestedTime'].fillna('2023-01-01')  # Fill with a placeholder date (or use latest date)
```

```python
# Checking for duplicates
ride_requests_duplicates = ride_requests_df.duplicated().sum()
ride_requests_duplicates
```

```python
# Checking for inconsistencies
ride_requests_inconsistencies = check_inconsistencies(ride_requests_df)
ride_requests_inconsistencies
```

```python
ride_requests_data_types_before = ride_requests_df.dtypes
ride_requests_data_types_before
```

```python
#fixing datatypes
ride_requests_df['RequestedTime'] = pd.to_datetime(ride_requests_df['RequestedTime'], errors='coerce')  # Convert RequestedTime to datetime
ride_requests_df['RequestID'] = ride_requests_df['RequestID'].astype('category')  # Convert RequestID to category
```

```python
# For categorical columns, using  mode to fill missing values
ride_requests_df['RequestedPickup'] = ride_requests_df['RequestedPickup'].fillna(ride_requests_df['RequestedPickup'].mode()[0])
ride_requests_df['RequestedDropoff'] = ride_requests_df['RequestedDropoff'].fillna(ride_requests_df['RequestedDropoff'].mode()[0])
```
