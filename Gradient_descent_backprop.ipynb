```python
import numpy as np
import matplotlib.pyplot as plt

# Define function and gradient
def J(theta):
    return theta**2

def gradient(theta):
    return 2 * theta

# Gradient Descent Parameters
alpha = 0.1  # Learning rate
theta = -4.0  # Initial guess
iterations = 10

# Store values for visualization
theta_values = [theta]
J_values = [J(theta)]

# Gradient Descent Iterations
for i in range(iterations):
    theta = theta - alpha * gradient(theta)  # Update rule
    theta_values.append(theta)
    J_values.append(J(theta))
    print(f"Iteration {i+1}: theta = {theta:.5f}, J(theta) = {J(theta):.5f}")

# Plot function and descent path
theta_plot = np.linspace(-6, 6, 100)
plt.plot(theta_plot, J(theta_plot), label="J(θ) = θ^2")
plt.scatter(theta_values, J_values, color="red", label="Gradient Descent Steps")
plt.xlabel("θ")
plt.ylabel("J(θ)")
plt.legend()
plt.title("Gradient Descent on J(θ) = θ^2")
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the convex function J(theta1, theta2)
def J(theta1, theta2):
    return theta1**2 + theta2**2  # Convex function

# Compute gradients
def gradient(theta1, theta2):
    dJ_dtheta1 = 2 * theta1
    dJ_dtheta2 = 2 * theta2
    return dJ_dtheta1, dJ_dtheta2

# Gradient Descent Parameters
alpha = 0.1  # Learning rate
theta1, theta2 = 2.0, -3.0  # Initial values
iterations = 20

# Store values for visualization
theta1_values = [theta1]
theta2_values = [theta2]
J_values = [J(theta1, theta2)]

# Gradient Descent Iterations
for i in range(iterations):
    grad_theta1, grad_theta2 = gradient(theta1, theta2)
    theta1 -= alpha * grad_theta1  # Update theta1
    theta2 -= alpha * grad_theta2  # Update theta2
    
    theta1_values.append(theta1)
    theta2_values.append(theta2)
    J_values.append(J(theta1, theta2))
    
    print(f"Iteration {i+1}: θ1 = {theta1:.5f}, θ2 = {theta2:.5f}, J(θ1,θ2) = {J(theta1, theta2):.5f}")

# Create a 3D plot of J(theta1, theta2)
theta1_range = np.linspace(-3, 3, 50)
theta2_range = np.linspace(-3, 3, 50)
theta1_mesh, theta2_mesh = np.meshgrid(theta1_range, theta2_range)
J_mesh = J(theta1_mesh, theta2_mesh)

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

# Plot the 3D surface
ax.plot_surface(theta1_mesh, theta2_mesh, J_mesh, cmap='viridis', alpha=0.7)

# Plot gradient descent path
ax.scatter(theta1_values, theta2_values, J_values, color='red', s=50, label="Gradient Descent Steps")

# Labels
ax.set_xlabel(r'$\theta_1$')
ax.set_ylabel(r'$\theta_2$')
ax.set_zlabel(r'$J(\theta_1, \theta_2)$')
ax.set_title("Gradient Descent on Convex Function $J(\Theta) = \Theta_1^2 + \Theta_2^2$")
ax.legend()

plt.show()
```

```python
import numpy as np
import pandas as pd

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def compute_loss(y, y_hat):
    return - (y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))

# Given data
x1, x2, y = 1, 2, 0
w1, w2, w3 = 0.2, 0.4, 0.3
b1, b2 = 1, 1
alpha = 0.01  # Learning rate
iterations = 5 # Number of iterations

# Store parameter updates
history = []

# Iterate over multiple updates
for i in range(iterations):
    # Forward pass
    z1 = w1*x1 + w2*x2 + b1
    o1 = sigmoid(z1)
    z2 = w3*o1 + b2
    y_hat = sigmoid(z2)
    loss = compute_loss(y, y_hat)
    
    # Store current parameters
    history.append([i+1, w1, w2, w3, b1, b2, loss])
    
    # Backpropagation
    # Gradient of loss w.r.t y_hat
    dL_dyhat = (y_hat - y)/(y_hat*(1-y_hat))
    # Gradient w.r.t z2
    dL_dz2 = dL_dyhat * y_hat * (1 - y_hat)
    # Gradient w.r.t w3 and b2
    dL_dw3 = dL_dz2 * o1
    dL_db2 = dL_dz2
    # Gradient w.r.t o1
    dL_do1 = dL_dz2 * w3
    # Gradient w.r.t z1
    dL_dz1 = dL_do1 * o1 * (1 - o1)
    # Gradient w.r.t w1, w2, and b1
    dL_dw1 = dL_dz1 * x1
    dL_dw2 = dL_dz1 * x2
    dL_db1 = dL_dz1
    
    # Update weights and biases
    w1 -= alpha * dL_dw1
    w2 -= alpha * dL_dw2
    w3 -= alpha * dL_dw3
    b1 -= alpha * dL_db1
    b2 -= alpha * dL_db2
    
    

# Store final results in a table
df = pd.DataFrame(history, columns=['Iteration', 'w1', 'w2', 'w3', 'b1', 'b2', 'Loss'])
print(df)
```
